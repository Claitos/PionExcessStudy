```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("TabulatedData/EntropyProfile_ALICE276Duke.txt")

# Example usage:
print("Shape:", data.shape)  # (rows, columns)
print("First column:", data[:, 0])
print("Second column:", data[:, 1])
```

```python
plt.figure(figsize=(8, 6))
for i in range(data.shape[1]-46):
    plt.plot(data[:, 0], data[:, i+1], label=f'Centrality {i}-{i+1} %' if i < 10 else "")

plt.title('Trento Profile (wo normalization) for ALICE Pb-Pb 276 GeV')
plt.xlabel('Radius (fm)')
plt.ylabel('Trento Profile')
plt.legend( loc='upper right', fontsize='small')
#plt.grid(True)
#plt.savefig('Plots/TrentoProfile_ALICE276.png', dpi=300)
plt.show()
```

```python
tau0 = 0.7 
norm = 35


plt.figure(figsize=(8, 6))
for i in range(data.shape[1]-46):
    plt.plot(data[:, 0], data[:, i+1]*norm/tau0, label=f'Centrality {i}-{i+1} %' if i < 10 else "")

plt.title('Entropy Profile (with normalization) for ALICE Pb-Pb 502 GeV')
plt.xlabel('Radius (fm)')
plt.ylabel('Entropy Density')
plt.legend( loc='upper right', fontsize='small')
#plt.grid(True)
#plt.savefig('Plots/EntropyProfile_ALICE276.png', dpi=300)
plt.show()
```
